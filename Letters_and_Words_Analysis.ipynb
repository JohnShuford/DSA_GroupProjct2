##### 1) Convert all text to lowercase

```python
with open('Little_Women.txt', 'r') as data_file:

    # open output.txt file in append mode
    with open('output.txt', 'a') as output_file:

        # read data.txt file, convert case,
        # and write to output.txt file
        output_file.write(data_file.read().lower())
```

##### 2) Remove punctuation, digits, and special characters (you may keep apostrophes inside words, specify your rule).

```python
import re

with open('output.txt', 'r') as file:
    text = file.read()

# Remove punctuation and special characters using regular expressions
clean_text = clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

# Print the cleaned text
print(clean_text)
```

##### 3) Tokenize by whitespace

```python
import re

text = clean_text
tokens = re.findall(r'\w+', text)
print(tokens)
```
