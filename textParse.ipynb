```python
from pypdf import PdfReader
```

```python
reader = PdfReader('Foundations_of_Data_ScienceTOC.pdf')
page = reader.pages
content = ['n', 'n', 'n', 'n', 'n']
for x in range(1, 4):
    page = reader.pages[x]
    content[x-1] = page.extract_text().split('\n')

full_text = [item for sublist in content for item in sublist]
```

```python
stripped = []
for item in full_text:
    newItem = item.replace(' .', '')
    stripped.append(newItem)
    
```
